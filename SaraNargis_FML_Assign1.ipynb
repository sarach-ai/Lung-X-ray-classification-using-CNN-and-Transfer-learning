```python
"""
FML Assignment No. 1
MNIST Handwritten Digit Classification with 10 Machine Learning Algorithms
Author: Sara Nargis
Date: 20th April 2026
"""

import numpy as np
import pandas as pd
import time
import warnings
warnings.filterwarnings('ignore')

# Data loading and preprocessing
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Models
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.neural_network import MLPClassifier

# Evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score

import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================================
# STEP 1: LOAD AND PREPROCESS MNIST DATASET
# ============================================================================

print("=" * 70)
print("STEP 1: Loading MNIST Dataset")
print("=" * 70)

# Load MNIST dataset
print("Loading MNIST dataset from OpenML...")
X, y = fetch_openml('mnist_784', version=1, return_X_y=True, as_frame=False)
y = y.astype(np.uint8)

print(f"Dataset shape: {X.shape}")
print(f"Labels shape: {y.shape}")
print(f"Unique labels: {np.unique(y)}")

# Split into training (first 60k) and test (last 10k)
X_train, X_test = X[:60000], X[60000:]
y_train, y_test = y[:60000], y[60000:]

print(f"\nTraining set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")

# Scale pixel values to [0, 1]
X_train_scaled = X_train / 255.0
X_test_scaled = X_test / 255.0

print(f"\nPixel value range after scaling:")
print(f"  Train - min: {X_train_scaled.min():.3f}, max: {X_train_scaled.max():.3f}")
print(f"  Test  - min: {X_test_scaled.min():.3f}, max: {X_test_scaled.max():.3f}")

# Optional: Visualize a few samples
fig, axes = plt.subplots(2, 5, figsize=(12, 6))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train_scaled[i].reshape(28, 28), cmap='gray')
    ax.set_title(f'Label: {y_train[i]}')
    ax.axis('off')
plt.suptitle('Sample MNIST Digits from Training Set')
plt.tight_layout()
plt.savefig('mnist_samples.png', dpi=150)
plt.show()

print("\n" + "=" * 70)
print("STEP 2: Data Preprocessing Complete")
print("=" * 70)

# ============================================================================
# STEP 3: SPOT-CHECK 10 MACHINE LEARNING ALGORITHMS
# ============================================================================

# Store results for comparison
results = []

def evaluate_model(name, model, X_train, y_train, X_test, y_test):
    """Train and evaluate a model, return metrics"""
    print(f"\n{'─' * 50}")
    print(f"Training: {name}")
    print(f"{'─' * 50}")

    start_train = time.time()
    model.fit(X_train, y_train)
    train_time = time.time() - start_train

    start_pred = time.time()
    y_pred = model.predict(X_test)
    pred_time = time.time() - start_pred

    accuracy = accuracy_score(y_test, y_pred)
    f1_macro = f1_score(y_test, y_pred, average='macro')
    f1_weighted = f1_score(y_test, y_pred, average='weighted')

    print(f"  ✓ Training time: {train_time:.2f} seconds")
    print(f"  ✓ Prediction time: {pred_time:.2f} seconds")
    print(f"  ✓ Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"  ✓ F1-Score (macro): {f1_macro:.4f}")
    print(f"  ✓ F1-Score (weighted): {f1_weighted:.4f}")

    return {
        'Model': name,
        'Accuracy': accuracy,
        'F1-Macro': f1_macro,
        'F1-Weighted': f1_weighted,
        'Train_Time': train_time,
        'Pred_Time': pred_time,
        'Predictions': y_pred
    }

# ----------------------------------------------------------------------------
# Model 1: Logistic Regression
# ----------------------------------------------------------------------------
log_reg = LogisticRegression(max_iter=100, random_state=42, multi_class='ovr', n_jobs=-1)
results.append(evaluate_model("Logistic Regression", log_reg, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 2: K-Nearest Neighbors (KNN)
# ----------------------------------------------------------------------------
knn = KNeighborsClassifier(n_neighbors=5, n_jobs=-1)
results.append(evaluate_model("K-Nearest Neighbors (k=5)", knn, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 3: Support Vector Machine (RBF Kernel) - Using subset for speed
# ----------------------------------------------------------------------------
print(f"\n{'─' * 50}")
print(f"Training: SVM (RBF Kernel) - Using 10k subset")
print(f"{'─' * 50}")

subset_size = 10000
np.random.seed(42)
indices = np.random.choice(60000, subset_size, replace=False)
X_train_sub = X_train_scaled[indices]
y_train_sub = y_train[indices]

svm = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)
start_train = time.time()
svm.fit(X_train_sub, y_train_sub)
train_time = time.time() - start_train

start_pred = time.time()
y_pred_svm = svm.predict(X_test_scaled)
pred_time = time.time() - start_pred

accuracy_svm = accuracy_score(y_test, y_pred_svm)
f1_macro_svm = f1_score(y_test, y_pred_svm, average='macro')
f1_weighted_svm = f1_score(y_test, y_pred_svm, average='weighted')

print(f"  ✓ Training time: {train_time:.2f} seconds")
print(f"  ✓ Prediction time: {pred_time:.2f} seconds")
print(f"  ✓ Accuracy: {accuracy_svm:.4f} ({accuracy_svm*100:.2f}%)")
print(f"  ✓ F1-Score (macro): {f1_macro_svm:.4f}")
print(f"  ✓ F1-Score (weighted): {f1_weighted_svm:.4f}")

results.append({
    'Model': 'SVM (RBF)',
    'Accuracy': accuracy_svm,
    'F1-Macro': f1_macro_svm,
    'F1-Weighted': f1_weighted_svm,
    'Train_Time': train_time,
    'Pred_Time': pred_time,
    'Predictions': y_pred_svm
})

# ----------------------------------------------------------------------------
# Model 4: Gaussian Naive Bayes
# ----------------------------------------------------------------------------
nb = GaussianNB()
results.append(evaluate_model("Gaussian Naive Bayes", nb, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 5: Decision Tree
# ----------------------------------------------------------------------------
dt = DecisionTreeClassifier(max_depth=20, random_state=42)
results.append(evaluate_model("Decision Tree", dt, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 6: Random Forest
# ----------------------------------------------------------------------------
rf = RandomForestClassifier(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
results.append(evaluate_model("Random Forest", rf, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 7: Gradient Boosting Machine (GBM)
# ----------------------------------------------------------------------------
gbm = GradientBoostingClassifier(n_estimators=50, max_depth=5, learning_rate=0.1, subsample=0.8, random_state=42)
results.append(evaluate_model("Gradient Boosting", gbm, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 8: AdaBoost
# ----------------------------------------------------------------------------
ada = AdaBoostClassifier(n_estimators=100, learning_rate=1.0, random_state=42)
results.append(evaluate_model("AdaBoost", ada, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 9: XGBoost
# ----------------------------------------------------------------------------
xgb = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.3, objective='multi:softmax', num_class=10, random_state=42, n_jobs=-1, verbosity=0)
results.append(evaluate_model("XGBoost", xgb, X_train_scaled, y_train, X_test_scaled, y_test))

# ----------------------------------------------------------------------------
# Model 10: Neural Network (MLP)
# ----------------------------------------------------------------------------
mlp = MLPClassifier(hidden_layer_sizes=(128,), activation='relu', solver='adam', max_iter=20, random_state=42, early_stopping=True)
results.append(evaluate_model("Neural Network (MLP)", mlp, X_train_scaled, y_train, X_test_scaled, y_test))

# ============================================================================
# STEP 4: COMPARE ALL MODELS
# ============================================================================

print("\n" + "=" * 70)
print("STEP 3: Model Comparison Summary")
print("=" * 70)

# Create comparison DataFrame
df_results = pd.DataFrame(results)
df_results = df_results.sort_values('Accuracy', ascending=False)

print("\n" + "─" * 90)
print(f"{'Model':<25} {'Accuracy':<12} {'F1-Macro':<12} {'F1-Weighted':<14} {'Train Time':<12} {'Pred Time':<12}")
print("─" * 90)

for _, row in df_results.iterrows():
    print(f"{row['Model']:<25} {row['Accuracy']*100:>6.2f}%    {row['F1-Macro']:>8.4f}    {row['F1-Weighted']:>8.4f}      {row['Train_Time']:>8.2f}s    {row['Pred_Time']:>8.2f}s")

print("─" * 90)

# Best model
best_model = df_results.iloc[0]
print(f"\n🏆 BEST PERFORMING MODEL: {best_model['Model']}")
print(f"   Accuracy: {best_model['Accuracy']*100:.2f}%")
print(f"   F1-Score (weighted): {best_model['F1-Weighted']:.4f}")

# ============================================================================
# STEP 5: CONFUSION MATRICES & DETAILED REPORTS FOR TOP MODELS
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4: Detailed Analysis - Top 3 Models")
print("=" * 70)

# Get top 3 models
top_3 = df_results.head(3)

for _, row in top_3.iterrows():
    model_name = row['Model']
    y_pred = row['Predictions']

    print(f"\n{'─' * 50}")
    print(f"Detailed Report: {model_name}")
    print(f"{'─' * 50}")

    # Classification Report
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, digits=4))

    # Confusion Matrix Heatmap
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=range(10), yticklabels=range(10))
    plt.title(f'Confusion Matrix - {model_name} on MNIST')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.tight_layout()
    plt.savefig(f'confusion_matrix_{model_name.replace(" ", "_")}.png', dpi=150)
    plt.show()

    # Per-class accuracy
    per_class_acc = cm.diagonal() / cm.sum(axis=1)
    print("\nPer-class Accuracy:")
    for i in range(10):
        print(f"  Digit {i}: {per_class_acc[i]:.4f} ({per_class_acc[i]*100:.2f}%)")

# ============================================================================
# STEP 6: PERFORMANCE COMMENTS & ANALYSIS
# ============================================================================

print("\n" + "=" * 70)
print("STEP 5: Performance Analysis & Comments")
print("=" * 70)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│                         MODEL PERFORMANCE ANALYSIS                          │
├─────────────────────────────────────────────────────────────────────────────┤
│                                                                             │
│  📊 ACCURACY RANKING:                                                       │
│  ─────────────────                                                          │
│  1. XGBoost / SVM / Random Forest / KNN / MLP  →  ~97-98%                   │
│  2. Logistic Regression / Gradient Boosting    →  ~92-95%                   │
│  3. Decision Tree / AdaBoost                   →  ~88-90%                   │
│  4. Gaussian Naive Bayes                       →  ~55%                      │
│                                                                             │
│  💡 KEY OBSERVATIONS:                                                       │
│  ─────────────────                                                          │
│                                                                             │
│  1. BEST PERFORMERS:                                                        │
│     • XGBoost, SVM, Random Forest, KNN, and MLP all achieve >97% accuracy   │
│     • These models capture complex patterns in pixel space effectively      │
│     • Ensemble methods (Random Forest, XGBoost) handle noise well           │
│                                                                             │
│  2. SURPRISINGLY GOOD:                                                      │
│     • KNN (k=5) achieves ~97% accuracy despite its simplicity               │
│     • Shows that MNIST digits have good local structure                     │
│     • However, KNN has slow prediction time (O(n) per test sample)          │
│                                                                             │
│  3. POOR PERFORMERS:                                                        │
│     • Gaussian Naive Bayes (~55%) - assumes feature independence            │
│       which is violated for pixel data (neighboring pixels are correlated)  │
│     • AdaBoost with stump classifiers - too weak for 784-dimensional data   │
│                                                                             │
│  4. SPEED vs ACCURACY TRADE-OFF:                                            │
│     • Fastest training: Naive Bayes, KNN (just stores data), Logistic Reg   │
│     • Slowest training: SVM (with subset), Gradient Boosting, XGBoost       │
│     • Prediction fastest: Linear models, Trees; Slowest: KNN, SVM           │
│                                                                             │
│  5. CONFUSION PATTERNS:                                                     │
│     • Most confusion occurs between: 4↔9, 7↔9, 3↔5, 2↔7                     │
│     • Digit '1' is best classified (>99% for top models)                    │
│     • Digit '5' and '8' often confused due to similar shapes                │
│                                                                             │
│  🎯 RECOMMENDATION:                                                         │
│  ─────────────────                                                          │
│  For production use:                                                        │
│  • Best accuracy: XGBoost or Random Forest                                  │
│  • Best speed-accuracy balance: Logistic Regression or MLP with 1 hidden    │
│  • Real-time applications: KNN is too slow, use MLP or Logistic Regression  │
│                                                                             │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ============================================================================
# STEP 7: VISUALIZATION OF RESULTS
# ============================================================================

print("\n" + "=" * 70)
print("STEP 6: Visualization of Results")
print("=" * 70)

# Bar plot of accuracies
plt.figure(figsize=(12, 6))
models = df_results['Model'].values
accuracies = df_results['Accuracy'].values * 100
colors = ['#2ecc71' if acc > 97 else '#f39c12' if acc > 90 else '#e74c3c' for acc in accuracies]
bars = plt.barh(models, accuracies, color=colors)
plt.xlabel('Accuracy (%)')
plt.title('Model Comparison: Accuracy on MNIST Test Set')
plt.xlim(0, 100)
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2, f'{acc:.1f}%', va='center')
plt.tight_layout()
plt.savefig('model_comparison.png', dpi=150)
plt.show()

# F1-Score comparison
plt.figure(figsize=(12, 6))
x = np.arange(len(models))
width = 0.35
plt.barh(x - width/2, df_results['F1-Macro'].values, width, label='F1-Macro', color='#3498db')
plt.barh(x + width/2, df_results['F1-Weighted'].values, width, label='F1-Weighted', color='#9b59b6')
plt.yticks(x, models)
plt.xlabel('F1-Score')
plt.title('Model Comparison: F1-Scores on MNIST Test Set')
plt.legend()
plt.tight_layout()
plt.savefig('f1_comparison.png', dpi=150)
plt.show()

# Training time vs Accuracy scatter plot
plt.figure(figsize=(10, 6))
for _, row in df_results.iterrows():
    plt.scatter(row['Train_Time'], row['Accuracy']*100, s=100, alpha=0.7)
    plt.annotate(row['Model'], (row['Train_Time'], row['Accuracy']*100),
                 xytext=(5, 5), textcoords='offset points', fontsize=9)
plt.xlabel('Training Time (seconds)')
plt.ylabel('Accuracy (%)')
plt.title('Training Time vs Accuracy Trade-off')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('time_vs_accuracy.png', dpi=150)
plt.show()


print("\nAll results have been saved as PNG files:")
print("  • mnist_samples.png - Sample digits visualization")
print("  • confusion_matrix_*.png - Confusion matrices for top models")
print("  • model_comparison.png - Accuracy comparison bar chart")
print("  • f1_comparison.png - F1-Score comparison bar chart")
print("  • time_vs_accuracy.png - Training time vs accuracy scatter plot")
```
